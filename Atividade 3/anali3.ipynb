```python
PixPorMunicipio = "https://olinda.bcb.gov.br/olinda/servico/Pix_DadosAbertos/versao/v1/odata/TransacoesPixPorMunicipio(DataBase=@DataBase)?@DataBase=''&$top=100&$format=json&$select=Estado,Regiao,VL_RecebedorPF"
ChavesPix = "https://olinda.bcb.gov.br/olinda/servico/Pix_DadosAbertos/versao/v1/odata/ChavesPix(Data=@Data)?@Data='2023-09-01'&$top=100&$format=json&$select=Nome,NaturezaUsuario,TipoChave"
```

```python
import pandas as pd
import requests
import json
```

```python
def requisicao_api(ppm,cp):
  resposta = requests.get(ppm)
  resposta2 = requests.get(cp)
  ##resposta3 = requests.get(cp)

  if(resposta.status_code == 200 and 
    resposta2.status_code == 200):
    dados = resposta.json()
    dados2 = resposta2.json()
         #sucesso na requisição
    dados = resposta.json()
    with open('Municipio.json', 'w', encoding='utf-8') as arquivo:   #salvar arquivo json
      json.dump(dados, arquivo, ensure_ascii=False, indent=4)
    with open('Chaves.json', 'w', encoding='utf-8') as arquivo:   #salvar arquivo json
      json.dump(dados2, arquivo, ensure_ascii=False, indent=4)
```

```python
requisicao_api(PixPorMunicipio,ChavesPix)
```

```python
Municipio = jsondata['value']
Chaves = jsondata2['value']
```

```python
muni = pd.json_normalize(Municipio)
chave = pd.json_normalize(Chaves)
```

```python
muni.to_csv("dados.csv", index=False, encoding="utf-8")
chave.to_csv("valores.csv", index=False, encoding="utf-8")
```
